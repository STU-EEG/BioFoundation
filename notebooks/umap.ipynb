```python
import h5py
import numpy as np

with h5py.File('embeddings_with_times.h5', 'r') as f:
    embeddings = f['embeddings'][:]  # Shape: (num_epochs, 1, 80, 869)
    time_starts_loaded = f['time_starts'][:]  # Shape: (num_epochs,)

print(f"Loaded embeddings shape: {embeddings.shape}")
print(f"Loaded time starts: {time_starts_loaded}")
```

```python
import umap
import matplotlib.pyplot as plt

# Convert list of arrays to a single array (194, 80, 869) -> (194, 80*869)
X = np.array([emb.squeeze() for emb in embeddings])  # shape: (194, 80, 869)
X = X.reshape(X.shape[0], -1)  # shape: (194, 69520)

# Reduce to 2D
reducer = umap.UMAP(n_components=2, random_state=42)
X_2d = reducer.fit_transform(X)

# Plot
plt.figure(figsize=(10, 8))
plt.scatter(X_2d[:, 0], X_2d[:, 1], s=50, alpha=0.7)
plt.title('UMAP Projection of Embeddings')
plt.xlabel('UMAP 1')
plt.ylabel('UMAP 2')
plt.show()
```
