```python
import numpy as np

data = np.load('untracked/embeddings_with_times.npz')
embeddings = data['embeddings']  # Shape: (num_epochs, 1, 80, 869)
time_starts_loaded = data['time_starts'][:]  # Shape: (num_epochs,)

print(f"Loaded embeddings shape: {embeddings.shape}")
print(f"Loaded time starts: {time_starts_loaded}")
```

```python
time_starts_loaded[110:142]
```

```python
import umap
import matplotlib.pyplot as plt

# Convert list of arrays to a single array (194, 80, 869) -> (194, 80*869)
X = np.array([emb.squeeze() for emb in embeddings])[110:142]  # shape: (194, 80, 869)
X = X.reshape(X.shape[0], -1)  # shape: (194, 69520)

reducer = umap.UMAP(n_components=2, random_state=42)
X_2d = reducer.fit_transform(X)
```

```python
plt.figure(figsize=(10, 8))
plt.scatter(X_2d[:, 0], X_2d[:, 1], s=50, alpha=0.7)

for i, time_start in enumerate(time_starts_loaded):
    plt.text(X_2d[i, 0], X_2d[i, 1], f'{time_start:.1f}s', fontsize=8, ha='right', va='bottom')


plt.title('UMAP Projection of Embeddings')
plt.xlabel('UMAP 1')
plt.ylabel('UMAP 2')
plt.show()
```

# dendrogram

```python
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram

Z = linkage(X_2d, method='ward')  # 'ward' minimizes variance; alternatives: 'single', 'complete', etc.


plt.figure(figsize=(20, 8))
dendrogram(Z, labels=[f'{t:.1f}s' for t in time_starts_loaded], leaf_rotation=90, leaf_font_size=8)
plt.title('Hierarchical Clustering Dendrogram of UMAP Points')
plt.xlabel('Time Starts (s)')
plt.ylabel('Distance')
plt.show()
```

```python
Z = linkage(X, method='ward')

plt.figure(figsize=(22, 8))
dendrogram(Z, labels=[f'{i}: {t:.1f}s' for i, t in enumerate(time_starts_loaded)], leaf_rotation=90, leaf_font_size=8)
plt.title('Hierarchical Clustering Dendrogram of Original Embeddings')
plt.xlabel('Epoch Index and Time Starts (s)')
plt.ylabel('Distance')
plt.show()
```
